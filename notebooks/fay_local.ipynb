# Fay: local notebook, remote execution (T07)
Kernel: `envs/serverless-v4/.venv`. Cells run on serverless through Databricks Connect; no data is stored locally.


```python
from databricks.connect import DatabricksSession
spark = DatabricksSession.builder.serverless(True).getOrCreate()
type(spark)
```

```python
df = spark.table("meridian.gold.payment_failure_daily")
df.count()  # runs remotely; look it up in system.query.history under your identity
```

```python
# Small results only: no toPandas() on large tables
df.groupBy("acquirer_id").sum("failed_payments").orderBy("acquirer_id").show()
```
